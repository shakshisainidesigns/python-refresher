```python
class employee:
    company = "HP"
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        
    def printDetails(self):
        print(f"the company of {self.name} is {self.company} with salary ${self.salary}")

    @staticmethod
    def printTime():
        print("the time is")

    @classmethod
    def printCompany(cls):
        print(f"the company is {cls.company}")
        
obj = employee("Shakshi", 50000)
obj.printDetails()
obj.printTime()
obj.printCompany()
```
